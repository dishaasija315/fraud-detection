# Exploratory Data Analysis (EDA)
Loading the PaySim dataset, taking a sample, and performing basic EDA.

```python
import pandas as pd
import numpy as np

# Load the dataset
file_path = '../data/PS_20174392719_1491204439457_log.csv'
df = pd.read_csv(file_path)

# Take a random sample of 500,000 rows
df_sample = df.sample(n=500000, random_state=42)

print(f"Original dataset shape: {df.shape}")
print(f"Sampled dataset shape: {df_sample.shape}")
```

```python
# 1. Columns and data types
print("\n--- Columns and Data Types ---")
print(df_sample.dtypes)
```

```python
# 2. Missing values
print("\n--- Missing Values ---")
print(df_sample.isnull().sum())
```

```python
# 3. Fraud rate
print("\n--- Fraud Rate ---")
# The target column in PaySim is 'isFraud'
fraud_rate = df_sample['isFraud'].mean() * 100
print(f"Fraud Rate: {fraud_rate:.4f}%")
```

```python
# 4. Fraud rate by transaction type
print("\n--- Fraud Rate by Transaction Type ---")
fraud_by_type = df_sample.groupby('type')['isFraud'].mean() * 100
print(fraud_by_type)
```

```python
# 5. Distribution of transaction amounts
print("\n--- Distribution of Transaction Amounts ---")
print(df_sample['amount'].describe())
```
